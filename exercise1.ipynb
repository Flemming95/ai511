## Task 3: String Slicing

Integer division finds the midpoint. The `[::-1]` slice reverses a sequence with a step of `-1`.

```python
def first_half(value):
    return value[:len(value) // 2]


def backward(value):
    return value[::-1]


print(first_half("Hello World!"))
print(backward("Hello World!"))
```

## Task 5: Pig Latin

Vowel-starting words receive `hay`; consonant-starting words move their first character and receive `ay`.

```python
def pig_latin(word):
    if not word:
        return word
    if word[0].lower() in "aeiou":
        return word + "hay"
    return word[1:] + word[0] + "ay"


print(pig_latin("apple"))
print(pig_latin("hello"))
```

## Task 6: Largest Palindrome

Check every product of two three-digit numbers and keep the largest product that reads the same backward.

```python
def palindrome():
    largest = 0
    for first in range(999, 99, -1):
        for second in range(first, 99, -1):
            product = first * second
            if product > largest and str(product) == str(product)[::-1]:
                largest = product
    return largest


print(palindrome())
```

## Task 7: Alternating Harmonic Series

A list comprehension creates exactly `n` terms, and `sum()` adds them.

```python
def alt_harmonic(n):
    terms = [(-1) ** (k + 1) / k for k in range(1, n + 1)]
    return sum(terms)


print(alt_harmonic(10))
```

## Task 8b: Inverse Association

A list comprehension selects the name whose associated value matches the input number.

```python
def inverse_association(number):
    associations = [("Marco", 1), ("Luca", 2), ("Alex", 3)]
    matches = [name for name, value in associations if value == number]
    return matches[0]


print(inverse_association(2))
```

## Task 9: Mutable and Immutable Objects

Mutating a list or set changes both names, while operations on immutable values create new objects. `id()` compares object identity.

```python
def mutable_vs_immutable():
    my_int = 10
    int_copy = my_int
    my_int += 1
    print("int:", int_copy == my_int, id(int_copy) == id(my_int))

    my_string = "hello"
    string_copy = my_string
    my_string += "!"
    print("str:", string_copy == my_string, id(string_copy) == id(my_string))

    my_list = [1, 2]
    list_copy = my_list
    my_list.append(3)
    print("list:", list_copy == my_list, id(list_copy) == id(my_list))

    my_tuple = (1, 2)
    tuple_copy = my_tuple
    my_tuple += (1,)
    print("tuple:", tuple_copy == my_tuple, id(tuple_copy) == id(my_tuple))

    my_set = {1, 2}
    set_copy = my_set
    my_set.add(3)
    print("set:", set_copy == my_set, id(set_copy) == id(my_set))


mutable_vs_immutable()
```

## Task 10: Custom Modules

```python
import calculator


def hypotenuse(first_side, second_side):
    first_square = calculator.multiply(first_side, first_side)
    second_square = calculator.multiply(second_side, second_side)
    sum_of_squares = calculator.add(first_square, second_square)
    return calculator.sqrt(sum_of_squares)


print(hypotenuse(3, 4))
```

## Task 12: Power Set

`itertools.combinations()` creates subsets of each possible size, and `chain` joins those groups into one iterator. Each tuple becomes a set.

```python
from itertools import chain, combinations


def power_set(iterable):
    elements = list(iterable)
    subsets = chain.from_iterable(
        combinations(elements, size) for size in range(len(elements) + 1)
    )
    return [set(subset) for subset in subsets]


print(power_set({1, 2, 3}))
```

## Task 16: Interruptible Random Walk

The `try` block performs the walk. If Ctrl+C raises `KeyboardInterrupt`, the current position is returned.

```python
from random import choice


def random_walk(max_iters=1e12):
    try:
        walk = 0
        directions = [1, -1]
        for _ in range(int(max_iters)):
            walk += choice(directions)
        return walk
    except KeyboardInterrupt:
        print("Walk interrupted by user.")
        return walk


print(random_walk(10))
```

## Task 19: Matrix Polynomial

The `@` operator computes matrix powers through multiplication instead of elementwise powers.

```python
import numpy as np


def matrix_polynomial():
    matrix_a = np.array([
        [3, 1, 4],
        [1, 5, 9],
        [-5, 3, 1],
    ])
    matrix_a_squared = matrix_a @ matrix_a
    matrix_a_cubed = matrix_a_squared @ matrix_a
    return -matrix_a_cubed + 9 * matrix_a_squared - 15 * matrix_a


print(matrix_polynomial())
```

## Task 20: Structured Array Creation

`ones()` creates the base array, `triu()` selects the upper triangle, and `tril()` selects the lower triangle. `k=1` starts one diagonal above the main diagonal.

```python
import numpy as np


def structured_matrix_product():
    ones = np.ones((7, 7), dtype=np.int64)
    matrix_a = np.triu(ones)
    matrix_b = np.tril(-ones) + np.triu(5 * ones, k=1)
    return (matrix_a @ matrix_b).astype(np.int64)


print(structured_matrix_product())
```

## Task 22: Block Matrix

`np.hstack()` joins blocks horizontally and `np.vstack()` joins complete block rows vertically. Zero-block shapes match the neighboring blocks.

```python
import numpy as np


def block_matrix():
    matrix_a = np.array([[0, 2, 4], [1, 3, 5]])
    matrix_b = np.array([
        [3, 0, 0],
        [3, 3, 0],
        [3, 3, 3],
    ])
    matrix_c = -2 * np.eye(3, dtype=int)
    identity = np.eye(3, dtype=int)

    top_row = np.hstack((np.zeros((3, 3), dtype=int), matrix_a.T, identity))
    middle_row = np.hstack((matrix_a, np.zeros((2, 2), dtype=int),
                            np.zeros((2, 3), dtype=int)))
    bottom_row = np.hstack((matrix_b, np.zeros((3, 2), dtype=int), matrix_c))
    return np.vstack((top_row, middle_row, bottom_row))


print(block_matrix())
```

## Task 23: Row-Stochastic Matrix

`sum(axis=1)` computes one sum per row. `keepdims=True` keeps the sums as a column so broadcasting normalizes each row.

```python
import numpy as np


def row_stochastic(matrix):
    matrix = np.asarray(matrix)
    row_sums = matrix.sum(axis=1, keepdims=True)
    return matrix / row_sums


print(row_stochastic(np.array([[1, 2], [3, 4]])))
```

## Task 24: Polynomial Approximation of Pi

`numpy.polynomial.Polynomial` represents a polynomial by its coefficient array. The arcsine Taylor series evaluated at `1/2` gives an approximation to pi after multiplication by 6.

```python
import math
import numpy as np
from numpy.polynomial import Polynomial


n_terms = 30
max_degree = 2 * (n_terms - 1) + 1
coeffs = np.zeros(max_degree + 1)

for n in range(n_terms):
    degree = 2 * n + 1
    coefficient = (math.factorial(2 * n)
                   / ((2 * n + 1) * math.factorial(n) ** 2 * 4 ** n))
    coeffs[degree] = coefficient

arcsin_poly = Polynomial(coeffs)
pi_approx = 6 * arcsin_poly(0.5)
is_close = np.allclose(pi_approx, np.pi)

print(f"Approximated Pi : {pi_approx}")
print(f"Actual Pi       : {np.pi}")
print(f"np.allclose     : {is_close}")
```
